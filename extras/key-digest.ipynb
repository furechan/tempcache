```python
import re
import pickle
import hashlib
```

# Key Digest\n\nExplores how cache keys are hashed into file-safe digest strings.\n`TempCache.key_digest()` pickles the key and hashes it with MD5, producing a 32-char hex string used as the cache filename.

```python
# A composite key is pickled then MD5-hashed to produce a deterministic, file-safe digest\ncomposite_key = ('my_function', (1, 2), {'param': 1.23})\n\ndigest = hashlib.md5(pickle.dumps(composite_key)).hexdigest()\ndigest
```

```python
# Digest is a valid hex string — safe to use as a filename\nre.fullmatch(r"^[0-9a-fA-F]+$", digest)
```

```python
# Sanity check: pattern with a non-hex char (W) should not match\nbool(re.fullmatch(r"^[0-9a-fA-F]+W$", digest))
```
